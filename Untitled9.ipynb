```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings, time
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
from sklearn.feature_selection import mutual_info_classif, RFE, VarianceThreshold
from sklearn.linear_model import LinearRegression, Ridge, Lasso, LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import (RandomForestClassifier, AdaBoostClassifier,
                              GradientBoostingClassifier)
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score,
                             accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, ConfusionMatrixDisplay,
                             roc_curve, roc_auc_score, classification_report,
                             silhouette_score, davies_bouldin_score)
from sklearn.multiclass import OneVsRestClassifier
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.stats import multivariate_normal

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (9, 5)
SEED = 42
np.random.seed(SEED)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# --- Load real UCI Blood Transfusion Service Center data ---
df = pd.read_csv('transfusion.data')

rename_map = {
    'Recency (months)': 'recency_months',
    'Frequency (times)': 'frequency',
    'Monetary (c.c. blood)': 'monetary_m1',
    'Time (months)': 'time_months',
    'whether he/she donated blood in March 2007': 'will_donate_next_campaign',
}
if 'Recency (months)' not in df.columns and len(df.columns) == 5:
    df.columns = ['recency_months', 'frequency', 'monetary_m1',
                  'time_months', 'will_donate_next_campaign']
else:
    df = df.rename(columns=rename_map)

df.insert(0, 'donor_id', [f'D{i:05d}' for i in range(len(df))])
df['will_donate_next_campaign'] = df['will_donate_next_campaign'].astype(int)

# --- Synthesize the two target columns that do not exist in UCI ---
rng = np.random.default_rng(SEED)
n = len(df)

# days_to_next_donation: driven by the real recency + frequency (same formula shape)
df['days_to_next_donation'] = np.clip(
    140 - 22 * np.log1p(df['frequency']) + 2.6 * df['recency_months']
    + rng.normal(0, 25, n), 5, 400
).round(1)

# next_donation_type: softmax over frequency (same formula shape as original)
s_wb  =  1.2 - 0.12 * df['frequency'].values
s_pl  = -0.4 + 0.16 * df['frequency'].values
s_plt = -0.8 + 0.13 * df['frequency'].values
S = np.vstack([s_wb, s_pl, s_plt]).T
S = np.exp(S - S.max(axis=1, keepdims=True))
P = S / S.sum(axis=1, keepdims=True)
u = rng.random(n)
cum = P.cumsum(axis=1)
idx = (u > cum[:, 0]).astype(int) + (u > cum[:, 1]).astype(int)
df['next_donation_type'] = np.array(['Whole Blood', 'Plasma', 'Platelets'])[idx]

print(df.shape)
df.head()
```

```python
print("Shape:", df.shape)
df.info()
print("\nMissing values:")
print(df.isna().sum()[df.isna().sum() > 0])
print("\nDuplicate rows:", df.duplicated().sum())
df.describe().T
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print("After de-duplication:", df.shape)

TARGETS = ['will_donate_next_campaign', 'days_to_next_donation', 'next_donation_type']

num_cols = [c for c in df.select_dtypes(include=np.number).columns if c not in TARGETS]
cat_cols = [c for c in df.select_dtypes(include='object').columns
            if c not in TARGETS + ['donor_id']]

# Only impute if there is actually something to impute
if len(num_cols) > 0 and df[num_cols].isna().any().any():
    df[num_cols] = df[num_cols].fillna(df[num_cols].median())

if len(cat_cols) > 0 and df[cat_cols].isna().any().any():
    df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])

print("Missing values remaining:", df.isna().sum().sum())
```

```python
outlier_cols = ['recency_months', 'frequency', 'monetary_m1',
                'time_months', 'days_to_next_donation']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, c in zip(axes.ravel(), outlier_cols):
    sns.boxplot(y=df[c], ax=ax, color='#e74c3c')
    ax.set_title(c)
axes.ravel()[-1].axis('off')       # hide unused subplot (we have 5 columns, 6 slots)
plt.suptitle('Before Outlier Treatment')
plt.tight_layout()
plt.show()
```

```python
def iqr_bounds(s, k=1.5):
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

outlier_report = {}
for c in outlier_cols:
    lo, hi = iqr_bounds(df[c])
    n_out = ((df[c] < lo) | (df[c] > hi)).sum()
    outlier_report[c] = n_out
    df[c] = df[c].clip(lo, hi)

for k, v in outlier_report.items():
    print(f"{k:25s} -> {v} capped")
```

```python
# UCI dataset has no categorical columns -> no encoding needed
df_enc = df.copy()
```

```python
FEATURES = [c for c in df_enc.columns if c not in TARGETS + ['donor_id']]
print("Number of features:", len(FEATURES))
print(FEATURES)

X       = df_enc[FEATURES].copy()
y_clf   = df_enc['will_donate_next_campaign']
y_reg   = df_enc['days_to_next_donation']
y_multi = df_enc['next_donation_type']
```

```python
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=FEATURES, index=X.index)
X_minmax = pd.DataFrame(MinMaxScaler().fit_transform(X), columns=FEATURES)

print("StandardScaler mean:", round(X_scaled.mean().mean(), 6),
      "std:", round(X_scaled.std().mean(), 4))
print("MinMaxScaler min:", X_minmax.min().min(), "max:", X_minmax.max().max())
X_scaled.head()
```

```python
vt = VarianceThreshold(threshold=0.01)
vt.fit(X_scaled)
low_var = [f for f, keep in zip(FEATURES, vt.get_support()) if not keep]
print("Low-variance features dropped:", low_var if low_var else "none")

# Mutual information
mi = mutual_info_classif(X_scaled, y_clf, random_state=SEED)
mi_s = pd.Series(mi, index=FEATURES).sort_values(ascending=False)
plt.figure(figsize=(9, 7))
mi_s.head(20).plot(kind='barh', color='#2980b9')
plt.gca().invert_yaxis()
plt.title('Top 20 Features by Mutual Information')
plt.xlabel('MI score')
plt.tight_layout()
plt.show()

# Random Forest importance
rf_imp = RandomForestClassifier(n_estimators=200, random_state=SEED).fit(X_scaled, y_clf)
imp_s = pd.Series(rf_imp.feature_importances_, index=FEATURES).sort_values(ascending=False)
plt.figure(figsize=(9, 7))
imp_s.head(20).plot(kind='barh', color='#27ae60')
plt.gca().invert_yaxis()
plt.title('Top 20 Features by Random Forest Importance')
plt.tight_layout()
plt.show()

# RFE (only 4 features available, so keep 3)
rfe = RFE(LogisticRegression(max_iter=2000), n_features_to_select=3).fit(X_scaled, y_clf)
selected_rfe = [f for f, s in zip(FEATURES, rfe.support_) if s]
print("\nRFE selected features:")
print(selected_rfe)
```

```python
# Classification split
X_tr, X_te, yclf_tr, yclf_te = train_test_split(
    X_scaled, y_clf, test_size=0.25, random_state=SEED, stratify=y_clf
)

# Regression split (aliases kept for PDF-compatible naming)
Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(
    X_scaled, y_reg, test_size=0.25, random_state=SEED
)
Xr_trr, Xr_tel, yr_trr, yr_tel = Xr_tr, Xr_te, yr_tr, yr_te
Xr_tc,  yr_tc                  = Xr_te, yr_te     # bias-variance evaluation split

# Multi-class split
Xm_tr, Xm_te, ym_tr, ym_te = train_test_split(
    X_scaled, y_multi, test_size=0.25, random_state=SEED, stratify=y_multi
)
```

```python
def reg_metrics(y_true, y_pred, n_feat, name):
    mse  = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae  = mean_absolute_error(y_true, y_pred)
    r2   = r2_score(y_true, y_pred)
    n    = len(y_true)
    adj  = 1 - (1 - r2) * (n - 1) / (n - n_feat - 1)
    return dict(Model=name, MSE=round(mse, 2), RMSE=round(rmse, 2),
                MAE=round(mae, 2), R2=round(r2, 4), Adj_R2=round(adj, 4))
```

```python
simple_fcat = 'recency_months'
slr      = LinearRegression().fit(Xr_trr[[simple_fcat]], yr_trr)
pred_slr = slr.predict(Xr_tel[[simple_fcat]])
res_simple = reg_metrics(yr_tel, pred_slr, 1, 'Simple Linear')
print(res_simple)
print("Intercept:", round(slr.intercept_, 2), "Slope:", round(slr.coef_[0], 2))

plt.figure(figsize=(8, 5))
plt.scatter(Xr_tel[simple_fcat], yr_tel, s=8, alpha=0.35, label='Actual')
order = np.argsort(Xr_tel[simple_fcat].values)
plt.plot(Xr_tel[simple_fcat].values[order], pred_slr[order],
         'r-', lw=2, label='Fitted')
plt.xlabel('recency_months (scaled)')
plt.ylabel('days_to_next_donation')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

```python
mlr      = LinearRegression().fit(Xr_trr, yr_trr)
pred_mlr = mlr.predict(Xr_tel)
res_multi = reg_metrics(yr_tel, pred_mlr, Xr_trr.shape[1], 'Multiple Linear')
print(res_multi)

coef_df = pd.Series(mlr.coef_, index=Xr_trr.columns).sort_values(key=abs, ascending=False)
print("\nTop 10 coefficients by magnitude:")
print(coef_df.head(10).round(3))
```

```python
poly_fcats = ['recency_months', 'frequency', 'time_months']
poly   = PolynomialFeatures(degree=2, include_bias=False)
Xp_tr  = poly.fit_transform(Xr_trr[poly_fcats])
Xp_tel = poly.transform(Xr_tel[poly_fcats])

plr      = LinearRegression().fit(Xp_tr, yr_trr)
pred_plr = plr.predict(Xp_tel)
res_poly = reg_metrics(yr_tel, pred_plr, Xp_tr.shape[1], 'Polynomial deg=2')
print("Polynomial feature count:", Xp_tr.shape[1])
print(res_poly)
```

```python
yscaler = StandardScaler()
yr_tr_s = yscaler.fit_transform(yr_tr.values.reshape(-1, 1)).ravel()

Xb_tr = np.c_[np.ones(len(Xr_trr)), Xr_trr.values]
Xb_te = np.c_[np.ones(len(Xr_tel)), Xr_tel.values]

def gradient_descent(Xb, y, lr=0.05, iters=800):
    theta = np.zeros(Xb.shape[1])
    costs = []
    m = len(y)
    for _ in range(iters):
        err  = Xb @ theta - y
        grad = (2 / m) * Xb.T @ err
        theta = theta - lr * grad
        costs.append(np.mean(err ** 2))
    return theta, costs

plt.figure(figsize=(9, 5))
for lr in [0.001, 0.01, 0.05, 0.2, 0.6]:
    _, c = gradient_descent(Xb_tr, yr_tr_s, lr=lr, iters=300)
    plt.plot(c, label=f'lr={lr}')
plt.yscale('log')
plt.xlabel('Iteration')
plt.ylabel('MSE (log scale)')
plt.title('Learning Rate Effect on Gradient Descent')
plt.legend()
plt.show()
```

```python
theta, costs = gradient_descent(Xb_tr, yr_tr_s, lr=0.05, iters=1500)
pred_gd_s = Xb_te @ theta
pred_gd   = yscaler.inverse_transform(pred_gd_s.reshape(-1, 1)).ravel()
res_gd    = reg_metrics(yr_tel, pred_gd, Xr_trr.shape[1], 'GD Linear Regression')
print("Final training cost:", round(costs[-1], 5))
print(res_gd)
```

```python
results_reg = [res_simple, res_multi, res_poly]

for name, model, params in [
    ('Ridge', Ridge(random_state=SEED),                    {'alpha': [0.01, 0.1, 1, 10, 100]}),
    ('Lasso', Lasso(random_state=SEED, max_iter=10000),    {'alpha': [0.001, 0.01, 0.1, 1, 10]}),
]:
    gs   = GridSearchCV(model, params, cv=5, scoring='r2', n_jobs=-1).fit(Xr_trr, yr_trr)
    pred = gs.best_estimator_.predict(Xr_tel)
    r    = reg_metrics(yr_tel, pred, Xr_trr.shape[1],
                       f'{name} alpha={gs.best_params_["alpha"]}')
    results_reg.append(r)
    print(name, "best alpha:", gs.best_params_['alpha'])
    if name == 'Lasso':
        n_zero = (np.abs(gs.best_estimator_.coef_) < 1e-8).sum()
        print("Lasso zeroed coefficients:", n_zero, "of", Xr_trr.shape[1])

reg_table = pd.DataFrame(results_reg)
print("\nRegression comparison:")
print(reg_table.to_string(index=False))
```

```python
train_scores, test_scores, degrees = [], [], range(1, 9)

for d in degrees:
    pf = PolynomialFeatures(degree=d, include_bias=False)
    Xt = pf.fit_transform(Xr_trr[poly_fcats])
    Xv = pf.transform(Xr_tc[poly_fcats])
    m  = LinearRegression().fit(Xt, yr_trr)
    train_scores.append(r2_score(yr_trr, m.predict(Xt)))
    test_scores.append(r2_score(yr_tc,  m.predict(Xv)))

plt.figure(figsize=(9, 5))
plt.plot(degrees, train_scores, 'o-', label='Train R2')
plt.plot(degrees, test_scores,  's-', label='Test R2')
plt.xlabel('Polynomial Degree')
plt.ylabel('R2')
plt.title('Bias-Variance Tradeoff')
plt.legend()
plt.show()
```

```python
def log_loss(y, p):
    p = np.clip(p, 1e-9, 1 - 1e-9)
    return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))

def logistic_gd(Xb, y, lr=0.1, iters=2000):
    theta  = np.zeros(Xb.shape[1])
    losses = []
    m = len(y)
    for _ in range(iters):
        p     = sigmoid(Xb @ theta)
        grad  = (1 / m) * Xb.T @ (p - y)
        theta = theta - lr * grad
        losses.append(log_loss(y, p))
    return theta, losses

Xb_lr = np.c_[np.ones(len(X_tr)), X_tr[['recency_months']].values]
_, losses_lr = logistic_gd(Xb_lr, yclf_tr.values, lr=0.1, iters=2000)
plt.figure(figsize=(9, 5))
plt.plot(losses_lr, color='#8e44ad')
plt.xlabel('Iteration')
plt.ylabel('Log loss')
plt.title('Logistic GD loss curve')
plt.show()
```

```python
def evaluate_clf(name, model, X_tr, y_tr, X_te, y_te):
    t0 = time.time()
    model.fit(X_tr, y_tr)
    elapsed = time.time() - t0
    pred = model.predict(X_te)
    prob = model.predict_proba(X_te)[:, 1]
    return dict(Model=name,
                Accuracy=round(accuracy_score(y_te, pred), 4),
                Precision=round(precision_score(y_te, pred), 4),
                Recall=round(recall_score(y_te, pred), 4),
                F1=round(f1_score(y_te, pred), 4),
                ROC_AUC=round(roc_auc_score(y_te, prob), 4),
                Time_s=round(elapsed, 2)), model, prob

models = {
    'Logistic Regression':    LogisticRegression(max_iter=2000, C=1.0, random_state=SEED),
    'Logistic L2 C=0.1':      LogisticRegression(max_iter=2000, C=0.1, penalty='l2', random_state=SEED),
    'Logistic L1 C=0.1':      LogisticRegression(max_iter=2000, C=0.1, penalty='l1',
                                                 solver='liblinear', random_state=SEED),
    'Decision Tree Gini':     DecisionTreeClassifier(criterion='gini', max_depth=6, random_state=SEED),
    'Decision Tree Entropy':  DecisionTreeClassifier(criterion='entropy', max_depth=6, random_state=SEED),
    'SVM RBF':                SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=SEED),
    'k-NN k=11':              KNeighborsClassifier(n_neighbors=11),
    'Naive Bayes':            GaussianNB(),
}

rows, fitted, probs = [], {}, {}
for name, m in models.items():
    r, fm, pr = evaluate_clf(name, m, X_tr, yclf_tr, X_te, yclf_te)
    rows.append(r)
    fitted[name] = fm
    probs[name]  = pr

clf_table = pd.DataFrame(rows).sort_values('ROC_AUC', ascending=False).reset_index(drop=True)
print("Classification results:")
print(clf_table.to_string(index=False))
```

```python
cluster_feats = ['recency_months', 'frequency', 'monetary_m1', 'time_months']
Xc = StandardScaler().fit_transform(df_enc[cluster_feats])

inertia, sil = [], []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=SEED).fit(Xc)
    inertia.append(km.inertia_)
    sil.append(silhouette_score(Xc, km.labels_))

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].plot(list(K_range), inertia, 'o-', color='#e67e22')
ax[0].axvline(4, ls='--', c='r')
ax[0].set_xlabel('K'); ax[0].set_ylabel('Inertia (WCSS)')
ax[0].set_title('Elbow Method')
ax[1].plot(list(K_range), sil, 's-', color='#16a085')
ax[1].axvline(4, ls='--', c='r')
ax[1].set_xlabel('K'); ax[1].set_ylabel('Silhouette Score')
ax[1].set_title('Silhouette Analysis')
plt.tight_layout()
plt.show()

K_OPT  = 4
kmeans = KMeans(n_clusters=K_OPT, n_init=10, random_state=SEED).fit(Xc)
df_enc['kmeans_cluster'] = kmeans.labels_
print("Chosen k:", K_OPT)
print("Silhouette:", round(silhouette_score(Xc, kmeans.labels_), 4))
print("Davies-Bouldin:", round(davies_bouldin_score(Xc, kmeans.labels_), 4))

profile = df_enc.groupby('kmeans_cluster')[
    cluster_feats + ['will_donate_next_campaign']
].mean().round(2)
profile['n_donors'] = df_enc['kmeans_cluster'].value_counts().sort_index()
print("\nCluster profile:")
print(profile)
```

```python
Xc_pca = PCA(n_components=2, random_state=SEED).fit_transform(Xc)

plt.figure(figsize=(9, 6))
palette = ['#e74c3c', '#3498db', '#2ecc71', '#f39c12']
for c in range(K_OPT):
    m = kmeans.labels_ == c
    plt.scatter(Xc_pca[m, 0], Xc_pca[m, 1], s=15, alpha=0.6,
                color=palette[c], label=f'Cluster {c}')

cent = PCA(n_components=2, random_state=SEED).fit(Xc).transform(kmeans.cluster_centers_)
plt.scatter(cent[:, 0], cent[:, 1], marker='X', s=250, c='black', label='Centroids')
plt.xlabel('PC1'); plt.ylabel('PC2')
plt.title('K-Means Donor Segments')
plt.legend()
plt.show()

def label_cluster(row):
    if row['frequency'] >= df_enc['frequency'].quantile(0.65) and row['recency_months'] <= 8:
        return 'Champion'
    if row['recency_months'] > 20 and row['frequency'] >= 3:
        return 'Lapsed'
    if row['frequency'] <= 2:
        return 'New'
    return 'Regular'

cluster_label_map    = profile.apply(label_cluster, axis=1).to_dict()
df_enc['segment']    = df_enc['kmeans_cluster'].map(cluster_label_map)
print(df_enc['segment'].value_counts())
```

```python
sample_idx = np.random.choice(len(Xc), min(250, len(Xc)), replace=False)
Z = linkage(Xc[sample_idx], method='ward')

plt.figure(figsize=(14, 5))
dendrogram(Z, truncate_mode='lastp', p=25, leaf_rotation=90, leaf_font_size=9)
plt.title('Hierarchical Clustering Dendrogram (Ward linkage, AGNES)')
plt.xlabel('Cluster size')
plt.ylabel('Distance')
plt.show()

agg = AgglomerativeClustering(n_clusters=K_OPT, linkage='ward').fit(Xc)
print("AGNES silhouette:", round(silhouette_score(Xc, agg.labels_), 4))

for link in ['single', 'complete', 'average', 'ward']:
    lab = AgglomerativeClustering(n_clusters=K_OPT, linkage=link).fit_predict(Xc)
    print(f"linkage = {link:9s}  silhouette = {silhouette_score(Xc, lab):.4f}")

db = DBSCAN(eps=0.9, min_samples=12).fit(Xc)
n_clusters = len(set(db.labels_)) - (1 if -1 in db.labels_ else 0)
n_noise    = (db.labels_ == -1).sum()
print(f"\nDBSCAN clusters: {n_clusters}, noise points: {n_noise} "
      f"({100 * n_noise / len(Xc):.1f}%)")

if n_clusters > 1:
    mask = db.labels_ != -1
    print("Silhouette excluding noise:",
          round(silhouette_score(Xc[mask], db.labels_[mask]), 4))

df_enc['dbscan_label'] = db.labels_
```

```python
anom_feats = ['recency_months', 'frequency', 'monetary_m1', 'time_months']
Xa = StandardScaler().fit_transform(df_enc[anom_feats])

# Univariate z-score
z_scores  = np.abs(Xa)
uni_flags = (z_scores > 3).any(axis=1)
print(f"Univariate Gaussian |z|>3: {uni_flags.sum()} anomalies ({100 * uni_flags.mean():.2f}%)")

# Multivariate Gaussian
mu   = Xa.mean(axis=0)
cov  = np.cov(Xa, rowvar=False)
dens = multivariate_normal(mean=mu, cov=cov, allow_singular=True).pdf(Xa)
thr  = np.percentile(dens, 1.0)
mv_flags = dens < thr
print(f"Multivariate Gaussian p<1%: {mv_flags.sum()} anomalies ({100 * mv_flags.mean():.2f}%)")

# Gaussian Mixture
gmm       = GaussianMixture(n_components=4, covariance_type='full', random_state=SEED).fit(Xa)
log_prob  = gmm.score_samples(Xa)
gmm_thr   = np.percentile(log_prob, 1.0)
gmm_flags = log_prob < gmm_thr
print(f"GMM low-likelihood 1%: {gmm_flags.sum()} anomalies ({100 * gmm_flags.mean():.2f}%)")

df_enc['anomaly_gmm']   = gmm_flags.astype(int)
df_enc['anomaly_score'] = -log_prob

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].hist(log_prob, bins=60, color='#3498db', alpha=0.8)
ax[0].axvline(gmm_thr, c='r', ls='--', lw=2, label='1% threshold')
ax[0].set_title('GMM Log-Likelihood Distribution')
ax[0].legend()

ax[1].scatter(Xc_pca[:, 0], Xc_pca[:, 1], c=gmm_flags, cmap='coolwarm', s=12, alpha=0.6)
ax[1].set_title('Anomalies in PCA Space')
ax[1].set_xlabel('PC1'); ax[1].set_ylabel('PC2')
plt.tight_layout()
plt.show()

print("\nSample anomalous donors:")
print(df_enc[gmm_flags][['recency_months', 'frequency', 'monetary_m1',
                          'time_months', 'anomaly_score']]
      .sort_values('anomaly_score', ascending=False).head(8).round(2))
```

```python
# Build the scored donor table (uses the winning ensemble = Random Forest)
scored = df_enc.copy()
scored['propensity'] = rf_final.predict_proba(X_scaled)[:, 1]

print("Best ensemble model:", ens_table.iloc[0]['Model'])
print("Scored donors:", scored.shape)


def recommend_donors(top_n=10, min_recency=3.0,
                     exclude_anomalies=True,
                     w_prop=0.55, w_freq=0.25, w_engage=0.20,
                     verbose=True):
    d = scored.copy()
    d = d[d['recency_months'] >= min_recency]

    if exclude_anomalies:
        d = d[d['anomaly_gmm'] == 0]

    if d.empty:
        print("No eligible donors match the filters.")
        return d

    d['s_prop']   = d['propensity']
    d['s_freq']   = (d['frequency'] - d['frequency'].min()) / \
                    (d['frequency'].max() - d['frequency'].min() + 1e-9)
    d['s_engage'] = 1 - (d['recency_months'] - d['recency_months'].min()) / \
                        (d['recency_months'].max() - d['recency_months'].min() + 1e-9)

    d['priority_score'] = (w_prop   * d['s_prop'] +
                           w_freq   * d['s_freq'] +
                           w_engage * d['s_engage']).round(4)

    d = d.sort_values('priority_score', ascending=False).head(top_n)

    cols = ['donor_id', 'recency_months', 'frequency', 'monetary_m1',
            'segment', 'propensity', 'priority_score']
    out = d[cols].reset_index(drop=True)
    out.index = out.index + 1

    if verbose:
        print(f"Top {top_n} recommended donors")
    return out


recommend_donors(top_n=10)
```
